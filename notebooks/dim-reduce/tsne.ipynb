# t-SNE: t-distributed Stochastic Neighbor Embedding

t-SNE is a non-linear dimensionality reduction technique used mostly for
visualization. It converts high-dimensional similarities between points into
joint probabilities, then finds a low-dimensional embedding (usually 2D) whose
pairwise similarities match as closely as possible. Points that are close in
the original space tend to stay close in the embedding, which makes clusters
visible.

Below we embed the sklearn digits dataset (1797 samples of 8x8 images, 64
features) into 2D and color each point by its digit label.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE

# Load the digits dataset: 1797 8x8 images -> 64 features each
digits = load_digits()
X, y = digits.data, digits.target
print(f"data shape: {X.shape}, labels: {sorted(set(y))}")

# Embed into 2D. Perplexity ~ number of effective nearest neighbors to preserve.
tsne = TSNE(n_components=2, perplexity=30, random_state=42)
X_embedded = tsne.fit_transform(X)
print(f"embedded shape: {X_embedded.shape}")
```

```python
plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_embedded[:, 0], X_embedded[:, 1], c=y, cmap="tab10", s=8, alpha=0.8)
plt.colorbar(scatter, label="digit")
plt.title("t-SNE embedding of the digits dataset")
plt.xlabel("t-SNE dimension 1")
plt.ylabel("t-SNE dimension 2")
plt.show()
```
